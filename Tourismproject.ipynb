## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Standardize column names
for name in datasets:
    datasets[name].columns = datasets[name].columns.str.strip()

# Remove duplicate rows
for name in datasets:
    datasets[name].drop_duplicates(inplace=True)
    datasets[name].reset_index(drop=True, inplace=True)

# Strip whitespace from text columns
for name, df in datasets.items():
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].astype(str).str.strip()

# Convert ID columns to string for safe merging
for name, df in datasets.items():
    for col in df.columns:
        if "Id" in col or col in ["TransactionId", "UserId", "VisitMode", "AttractionId"]:
            df[col] = df[col].astype(str)
````

### What all manipulations have you done and insights you found?

The data wrangling process included standardizing column names, removing duplicate rows, trimming whitespace from text values, and converting identifier columns to a consistent string format. These changes make the data easier to merge and reduce the risk of mismatched keys during joins. The tables appear well-structured and suitable for combining into a master dataset for analysis. The next step is to validate relationships between tables and build a consolidated dataset for visualization and machine learning.

````python
# Chart - 10 visualization code
# Combined chart: Total Visits (bars) + Average Rating (line) by Month

# Prepare data
monthly_visits = df["VisitMonth"].value_counts().sort_index()
monthly_rating = df.groupby("VisitMonth")["Rating"].mean().sort_index()

months = monthly_visits.index

fig, ax1 = plt.subplots(figsize=(12, 6))

# Bar plot for total visits (left y-axis)
bars = ax1.bar(
    months,
    monthly_visits.values,
    color="skyblue",
    alpha=0.8,
    label="Total Visits"
)
ax1.set_xlabel("Visit Month", fontsize=12)
ax1.set_ylabel("Number of Visits", fontsize=12, color="blue")
ax1.tick_params(axis="y", labelcolor="blue")
ax1.set_xticks(range(1, 13))

# Add labels on bars
for bar in bars:
    height = bar.get_height()
    ax1.text(
        bar.get_x() + bar.get_width() / 2,
        height + (max(monthly_visits.values) * 0.01),
        str(int(height)),
        ha="center",
        va="bottom",
        fontsize=9
    )

# Secondary axis for average rating (right y-axis)
ax2 = ax1.twinx()
ax2.plot(
    months,
    monthly_rating.values,
    color="purple",
    marker="o",
    linewidth=2.5,
    label="Average Rating"
)
ax2.set_ylabel("Average Rating", fontsize=12, color="purple")
ax2.tick_params(axis="y", labelcolor="purple")
ax2.set_ylim(0, 5.5)

# Add labels on line points
for x, y in zip(months, monthly_rating.values):
    ax2.text(x, y + 0.05, f"{y:.2f}", ha="center", fontsize=9, color="purple")

# Title and grid
plt.title("Visits and Average Rating by Month", fontsize=16, fontweight="bold")
ax1.grid(True, linestyle="--", alpha=0.4)

# Legend
lines_labels = [ax1.get_legend_handles_labels(), ax2.get_legend_handles_labels()]
handles = lines_labels[0][0] + lines_labels[1][0]
labels = lines_labels[0][1] + lines_labels[1][1]
plt.legend(handles, labels, loc="upper right")

plt.tight_layout()
plt.show()
````

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import f_oneway

# Prepare groups
visitmode_groups = [group["Rating"].dropna().values for name, group in df.groupby("VisitMode")]

# One-way ANOVA
f_stat, p_value = f_oneway(*visitmode_groups)

print("ANOVA F-statistic:", f_stat)
print("P-value:", p_value)
````

````python
# Handling Outliers & Outlier treatments
import numpy as np

# Example: Handle outliers in Rating using IQR
if "Rating" in df.columns:
    Q1 = df["Rating"].quantile(0.25)
    Q3 = df["Rating"].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df["Rating"] = np.where(df["Rating"] < lower_bound, lower_bound, df["Rating"])
    df["Rating"] = np.where(df["Rating"] > upper_bound, upper_bound, df["Rating"])
````

````python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib

# A. Load models and small data (saved earlier in Colab)

@st.cache_resource
def load_models():
    """
    Load the pre-trained classification and regression models
    from the pkl files saved in Colab.
    """
    clf_model = joblib.load("visitmode_model.pkl")
    reg_model = joblib.load("rating_model.pkl")
    return clf_model, reg_model

@st.cache_data
def load_data():
    """
    Load the small recommendation dataframe with user,
    attraction, visit mode, and rating information.
    """
    df_ml_small = pd.read_csv(
        "df_ml_small.csv",
        dtype={"UserId": str, "AttractionId": str}
    )
    return df_ml_small

clf_model, reg_model = load_models()
df_ml_small = load_data()

# B. Recommendation helper functions

@st.cache_data
def build_attraction_stats(df):
    """
    Compute average rating and number of visits for each attraction,
    then keep only popular attractions (e.g., at least 10 visits),
    sorted by rating and popularity.
    """
    attraction_stats = (
        df.groupby("AttractionId")
        .agg(
            avg_rating=("Rating", "mean"),
            num_visits=("Rating", "count")
        )
        .reset_index()
    )

    popular_attractions = attraction_stats[
        attraction_stats["num_visits"] >= 10
    ].sort_values(
        by=["avg_rating", "num_visits"],
        ascending=False
    )

    return attraction_stats, popular_attractions

attraction_stats, popular_attractions = build_attraction_stats(df_ml_small)

def recommend_top_n_global(n=5):
    """
    Return the global top-N attractions by rating and visit count.
    """
    return popular_attractions.head(n)

def recommend_for_user(user_id, n=5):
    """
    Recommend attractions for a specific user:

    1. Find the user's main visit mode (e.g., Family, Solo).
    2. Find popular attractions for that visit mode.
    3. Exclude attractions the user has already visited.
    4. Return the top-N remaining attractions.
    """
    user_id = str(user_id)
    user_hist = df_ml_small[df_ml_small["UserId"] == user_id]

    if user_hist.empty:
        return recommend_top_n_global(n)

    main_mode = user_hist["VisitMode"].mode()[0]

    mode_stats = (
        df_ml_small[df_ml_small["VisitMode"] == main_mode]
        .groupby("AttractionId")
        .agg(
            avg_rating=("Rating", "mean"),
            num_visits=("Rating", "count")
        )
        .reset_index()
    )

    mode_stats = mode_stats[
        mode_stats["num_visits"] >= 5
    ].sort_values(
        by=["avg_rating", "num_visits"],
        ascending=False
    )

    visited = set(user_hist["AttractionId"].unique())
    mode_stats = mode_stats[~mode_stats["AttractionId"].isin(visited)]

    if mode_stats.empty:
        return recommend_top_n_global(n)

    return mode_stats.head(n)

@st.cache_data
def build_monthly_stats(df):
    """
    Build monthly average rating and visit count for charts.
    Assumes VisitMonth is present in df_ml_small or can be derived.
    """
    # If VisitMonth is not in df_ml_small, this will fail;
    # but in your project df_ml_small came from df_ml, which has VisitMonth.
    if "VisitMonth" not in df.columns:
        return None

    monthly = (
        df.groupby("VisitMonth")
        .agg(
            avg_rating=("Rating", "mean"),
            num_visits=("Rating", "count")
        )
        .reset_index()
        .sort_values("VisitMonth")
    )
    return monthly

# Try to attach VisitMonth if it exists in original df_ml_small
monthly_stats = build_monthly_stats(df_ml_small) if "VisitMonth" in df_ml_small.columns else None

# C. Streamlit user interface

st.title("Tourism Intelligence App")
st.caption("Rating Prediction, Visit Mode Classification, and Attraction Recommendation")

# Sidebar
st.sidebar.header("User & Visit Inputs")
user_id = st.sidebar.text_input(
    "UserId (for recommendations)",
    value=df_ml_small["UserId"].iloc[0]
)
visit_year = st.sidebar.number_input(
    "Visit Year", min_value=2015, max_value=2030, value=2022
)
visit_month = st.sidebar.number_input(
    "Visit Month", min_value=1, max_value=12, value=7
)

# Tabs for better organization
tab_pred, tab_reco, tab_viz = st.tabs(
    ["1. Prediction", "2. Recommendations", "3. Charts"]
)

# ---------- Tab 1: Prediction ----------
with tab_pred:
    st.subheader("Visit Mode Classification & Rating Prediction")
    st.write("Click **Predict** to get visit mode and rating predictions for a new visit.")

    if st.button("Predict"):
        # Base features from user input
        input_df = pd.DataFrame({
            "VisitYear": [visit_year],
            "VisitMonth": [visit_month],
        })

        # Add all other columns the model expects, with simple default values.
        # Adjust these defaults based on your dataset if needed.

        # IDs and basic categorical fields
        input_df["CityId"] = "0"
        input_df["CountryId"] = "0"
        input_df["RegionId"] = "0"
        input_df["ContinentId"] = "0"
        input_df["RegionId_country"] = "0"
        input_df["ContinentId_region"] = "0"

        # Use an existing AttractionId from the data as a placeholder
        input_df["AttractionId"] = df_ml_small["AttractionId"].iloc[0]
        input_df["AttractionTypeId"] = "0"
        input_df["AttractionType"] = "Unknown"

        # Time / derived fields
        input_df["VisitYearMonth"] = (
            input_df["VisitYear"].astype(str) + "-" + input_df["VisitMonth"].astype(str)
        )
        input_df["VisitSeason"] = "Unknown"

        # Text features for attraction (neutral strings)
        input_df["Attraction"] = "Unknown attraction"
        input_df["AttractionAddress"] = "Unknown address"
        input_df["Attraction_lower"] = input_df["Attraction"].str.lower()
        input_df["Attraction_no_punct"] = input_df["Attraction_lower"]
        input_df["Attraction_no_stop"] = input_df["Attraction_lower"]
        input_df["Attraction_no_url_digit"] = input_df["Attraction_lower"]
        input_df["Attraction_lemmas"] = input_df["Attraction_lower"]
        input_df["Attraction_pos"] = input_df["Attraction_lower"]
        input_df["Attraction_tokens"] = input_df["Attraction_lower"]

        # Location names
        input_df["AttractionCityId"] = "0"
        input_df["City"] = "Unknown city"
        input_df["Country"] = "Unknown country"
        input_df["Region"] = "Unknown region"
        input_df["Continent"] = "Unknown continent"

        # Model predictions
        try:
            pred_mode = clf_model.predict(input_df)[0]
            pred_rating = reg_model.predict(input_df)[0]

            st.markdown("### Predicted Visit Mode")
            st.write(pred_mode)

            st.markdown("### Predicted Rating")
            st.write(round(float(pred_rating), 2))
        except Exception as e:
            st.error(f"Prediction error: {e}")

# ---------- Tab 2: Recommendations ----------
with tab_reco:
    st.subheader("Attraction Recommendations")

    col1, col2 = st.columns(2)

    with col1:
        st.markdown("#### Global Top Attractions")
        top_global = recommend_top_n_global(5)
        st.dataframe(top_global)

    with col2:
        st.markdown("#### Personalized Recommendations")
        rec_user = recommend_for_user(user_id, 5)
        st.write(f"Recommendations for UserId: {user_id}")
        st.dataframe(rec_user)

# ---------- Tab 3: Charts ----------
with tab_viz:
    st.subheader("Overview Charts")

    # Bar chart: Top 10 attractions by average rating
    st.markdown("#### Top Attractions by Average Rating")
    top10 = attraction_stats.sort_values(
        "avg_rating", ascending=False
    ).head(10)
    st.bar_chart(
        data=top10.set_index("AttractionId")["avg_rating"],
        use_container_width=True,
    )

    # Line chart: Monthly average rating (if available)
    if monthly_stats is not None:
        st.markdown("#### Monthly Average Rating")
        st.line_chart(
            data=monthly_stats.set_index("VisitMonth")["avg_rating"],
            use_container_width=True,
        )
    else:
        st.info("Monthly statistics are not available because VisitMonth is missing in df_ml_small.")
````
